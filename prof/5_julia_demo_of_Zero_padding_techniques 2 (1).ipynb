Code written in Julia 

```julia

using FFTW  

dt=0.04
t=collect(0:dt:1);   # Create an array holding the time values

N=length(t);
println("Number of samples N = ",N)

f0 = 3.0  # Hz sinusoid

x = cos.(2*pi*f0*t);

X = fft(x);

using Plots
Plots.plotly();    # Specify Plotly backend which allows zooming with a mouse in Jupyter Notebook.
Plots.default(size=(800,300)); #Plot canvas size
Plots.default(label=""); # Turn off legends
Plots.default(ticks=:native);  #Ticks on x-axis are labelled nicely when zooming in.
Plots.default(grid=:false);  #Turn off grid
Plots.default(markersize=3)
Plots.default(linewidth=2)

fig = plot(x, markershape = :circle)

title!("Sampled signal x")
display(fig)


fig = plot(abs.(X), markershape = :circle)
title!("abs.( fft(x) )")
display(fig)


fig = plot(fftshift(abs.(X)), markershape = :circle)
title!("fftshift( abs.( fft(x) ) )")
display(fig)


# One needs to zoom in to see detail
# fig = plot(abs.(X[1:10]), markershape = :circle)
# title!("Zoomed in on left side")
# display(fig)


println("Maximum value = ", maximum(abs.(X)))
```

## Now use zero-padding to increase the number of elements in the signal
This will provide more samples in the frequency domain

```julia

N = length(x);

# Create array of zeros of length 10*N
# and copy array x into first N elements of y

y = zeros(10*N);
y[1:N] = x;    


fig = plot(y, markershape = :circle)
title!("Array extended by zero-padding to have 10x as many samples")
display(fig)

Y = fft(y);

fig = plot(abs.(Y), markershape = :circle)
title!("FFT of zero padded version of x")
display(fig)

fig = plot(fftshift(abs.(Y)), markershape = :circle)
title!("fftshift of FFT of zero padded version of x")
display(fig)

```

## Zoom in to show detail


```julia
zoom_range = 50:200
fig = plot(zoom_range,fftshift(abs.(Y))[zoom_range], markershape = :circle)
title!("Zoom in near the centre")
display(fig)

fig = plot(abs.(Y[1:length(zoom_range)]), markershape = :circle)
display(fig)

# FFT as an estimate of the Fourier Transform (FT)
Yscaled = Y*dt;   # Multiply by dt to obtain same amplitides as a FT

# In the case of cos(ωt)*rect( (t-Tp/2)/Tp )
# From FT, theory, the FT contains a pair of Sa() functions of peak amplitude Tp/2
# In this case Tp = N*dt;

Tp = N*dt
println("Pulse length Tp = ",Tp)
println("theoretical_peak_amplitude_of_FT = ",Tp/2)


fig = plot(abs.(Yscaled), markershape = :circle); 
title!(string("DFT approximation of an FT of cos(ωt)*rect( (t-Tp/2)/Tp where Tp= ",Tp))
display(fig)

```

## Example of Zero Padding in the frequency domain


```julia


dt=0.04
t=collect(0:dt:1);   # Create an array holding the time values

N=length(t);
println("Number of samples N = ",N)

f0 = 3.0  # Hz sinusoid

x = cos.(2*pi*f0*t);

fig = plot(t,x, markershape = :circle)
title!("Sampled signal x")
display(fig)

X = fft(x);

fig = plot(abs.(X), markershape = :circle)
title!("abs.( fft(x) )")
display(fig)


# Do zero-pad the DFT array by inserting 9*N zeros at the centre

if mod(N,2)==0   # Case N even
    k_mid = Int(N/2)
else
    k_mid = Int((N+1)/2)  # Case N odd
end


Y = zeros(10*N)+im*zeros(10*N)     # Create an array of zeros and copy in the correct parts of X
Ny = length(Y)
Y[1:k_mid]=X[1:k_mid]
Y[Ny-k_mid+1:Ny]=X[k_mid+1:N]

# Alternative code:   Y = [ X[1:k_mid] zeros(9*N) X[k_mid+1:N]]

fig = plot(abs.(Y), markershape = :circle)
title!("Zero-padded version of X, 10x longer")
display(fig)

# Now go back to time domain

y = ifft(Y)

fig = plot(real.(y), markershape = :circle)
title!("After inverse transforming the zero-padded FFT")
display(fig)

println("Ny = $Ny")
```

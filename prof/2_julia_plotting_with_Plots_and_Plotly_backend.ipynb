## Plotting with Plots + Plotyly backend


```julia
# Plots + Plotyly backend
# Plots defines a standard set of plot functions, and calls a backend plotting library to create the plot.
# To specify which backend to use, execute a function call, either gr(), plotly() or pythonplot(). 
# Default backend is gr() [using the GR library]. 
# The plotly() backend allows zooming with a mouse in Jupyter Notebook.
# The pythonplot() backend uses PythonPlot (Python's matplotlib), which does not allow mouse zooming in Jupyter notebook. 
#     PythonPlot allows zooming with the mouse if used from the REPL command line.
# Note: The initialisation time is irritatingly slow (as long as 1 minute), but thereafter it is quick to plot.
# Note2: In the code below, functions can be called via Plots.func() e.g. Plots.plot(), rather than just plot().  
# This allows one to use several plotting libraries at the same time. e.g.  One can also do PythonPlot.plot()

using Plots
plotly();    # Specify Plotly backend which allows zooming with a mouse inside Jupyter Notebook.

default(size=(800,300)); #Plot canvas size

default(label=""); # Turn off legends

default(ticks=:native);  #Ticks on x-axis are labelled nicely when zooming in.
   # ticks=:native Tells backend to calculate ticks by itself.
   # Good idea if you use interactive backends like plotly where you perform mouse zooming

# default(grid=:false);  #Turn off grid
# default(linewidth=2);  #Set line width for plot lines

# Specify font and canvas size defaults if you wish 
# fntsm = font("sans-serif", pointsize=round(10.0))
# fntlg = font("sans-serif", pointsize=round(14.0))
# default(titlefont=fntlg, guidefont=fntlg, tickfont=fntsm, legendfont=fntsm)

# It can be useful to change just the title font size via keyword:  titlefontsize=10

# Description of all adjustable Plots parameters:https://docs.juliaplots.org/latest/generated/attributes_series/
```

```julia
# Do some simple plotting  (try to zoom with your mouse's left button)
# This first plot can take several seconds to appear; thereafter it is instant.
t=range(0,10.0,step=0.1);  
y=cos.(t);  
fig1 = plot(t,y, label = "");
display(fig1);  # Display it now.
```

```julia
# Add another plot on same axes using plot!()
z=sin.(t);  
plot!(t,z, title = "Two Sinusoids on same axes", label = "Line 2", linewidth = 3);
xlabel!("time [s]");  # One can also add pamaeters afterwards (use the "!" to add)
display(fig1);
```

```julia
# Save a plot to a png file in the same folder as the notebook
# If this does not work, you will have to add a package: using Pkg; Pkg.add("PlotlyBase") 
# Maybe this does not work with all backends....(check!)
# savefig(fig1, "fig1.png");
```

```julia
# A plot command with more parameters that you can adjust (e.g. smaller size and no "key" box):
fig2 = plot(t, z, linestyle=:solid, linealpha=0.5, linewidth=4, linecolor=:red, size=(600,300), label = "", title = "Sinusoid", xlabel = "time [s]", ylabel = "z");
display(fig2);
```

```julia
# Scatter plot (using the scatter command)
scatter(t,z, markersize=2)
```

```julia
# Scatter plot (using the plot command, specifying series type)
fig3 = plot(t,z, seriestype = :scatter, markersize=2, title = "Scatter Plot showing markers (dots)");
display(fig3);
```

```julia
# A plot with both lines and markers
fig4 = plot( t,z, markershape = :circle, markersize = 2, markercolor=:lightblue, markerstrokecolor=:red)  
title!("Magnitude of FFT(v)")
display(fig4)
```

```julia
# A "stem" plot with markers
fig5=plot(t,z, line=:stem, marker=:circle, markersize=2)
display(fig5)
```

```julia
# Histogram

y = randn(100000)

fig6 = histogram(y, nbins = 20);
display(fig6)
```

```julia
# Example of displaying multiple plots in the same window

t=range(0,10.0,step=0.1);  
y=cos.(t);  
z=sin.(t);  

fig1 = plot(t,y, label = "Line 1");
fig2 = plot(t, z, linestyle=:solid, linealpha=0.5, linewidth=4, linecolor=:red, size=(600,300), label = "")
fig3 = plot(t,z, seriestype = :scatter, markersize=1);
fig4 = histogram(randn(1000), nbins = 20);

# The statement below combines all four figures into a single figure (fig5).

fig5 = plot(fig1, fig2, fig3, fig4, layout = (2, 2), legend = false)

display(fig5)
```

```julia
N=10
x = 0:N-1
y = x + 2.0*randn(N)
fig = scatter(x,y)
display(fig)

fig=scatter(x,y, smooth=true)  # Adds a linear regression line to your plots
display(fig)
```

```julia
# One can also do surface, contour plots and heatmaps of functions of two variables - see Plots documentation for details.
# Use the mouse to rotate the surface plot.
# The plot/surfuace/countour/heatmap function will evaluate function f at all specified values f(x,y)

x=-1:0.1:1
y=-1:0.1:1
f(x,y) = x^2 + y^2    # Define a function to plot

display( surface(x,y,f) )

display( contour( x,y,f , levels=10, color=:turbo, clabels=true, cbar=false, lw=1 ) )

# One can get plotly to show the correct aspect ratio by redefining the canvas size
display( contour( x,y,f , levels=10, color=:turbo, clabels=true, cbar=false, lw=1, size=(300*length(x)/length(y),300)) )


display( heatmap(x,y,f) )

# One can get plotly to show the correct aspect ratio by redefining the canvas size 
# (allowing a bit extra (+70 in this case) on the x dimension for the colorbar on the RHS):
display( heatmap(x,y,f, size=(300*length(x)/length(y)+70,300) ) )  
```

```julia
# Example of displaying a heatmap image of a matrix (with correct aspect ratio)
Nx=40
Ny=40
matrix = abs.(randn(Ny,Nx));  # Fill a 2D array with some non-negative random numbers   

display( heatmap(matrix, aspect_ratio=:equal) )   # Display a “heatmap” image with aspect_ratio keyword

# For some reason, plotly() currently does not seem to support keyword aspect_ratio, so termporarily switch to gr()
gr()
display( heatmap(matrix, aspect_ratio=:equal, title="Using gr() with apect_ratio=:equal" ) )  # Display a “heatmap” image with aspect_ratio keyword
plotly()
;

# One can get plotly to show the correct aspect ratio by redefining the canvas size, and allowing a bit extra (+55 approximately) on the x dimension for the colorbar on the RHS):
display( heatmap(matrix, size=(500*Nx/Ny+55,500), titlefontsize=11, title="heatmap(matrix, size=(300*Nx/Ny+55,300), titlefontsize=6 )" ))
```

```julia
# Plotting complex functions using 3D plot function
# A complex function of time or frequency can be visualised by plots of: real, imag, mag or phase.
# One can also illustrate the the function as a locus in 3D
j = im
ω0 = 1
T = 2*π/ω0
t = 0:T/100:4*T

g = exp.(j*ω0*t)   # Plot this complex exponential as a function of time

# display( plot( t, real(g), xlabel="time t", ylabel = "real" ))
# display( plot( t, imag(g), xlabel="time t", ylabel = "imag" ))
# display( plot( t, abs.(g), xlabel="time t", ylabel = "mag" ))
# display( plot( t, angle.(g), xlabel="time t", ylabel = "phase" ))

p = plot3d((t,real(g),imag(g)),linewidth=2, linecolor=:blue, size=(600,600), label = "")

xaxis!("time t",(0,maximum(t)),0:5:maximum(t)); 
yaxis!("real",(-2,2),-1:0.25:1); 
zaxis!("imag",(-2,2),-1:0.25:1); 

# Also draw red lines for the three axes
plot3d!( ([0; maximum(t)],[0; 0],[0; 0]), linewidth=4, linecolor=:red, label = "")
plot3d!( ([0; 0],[-1; 1],[0; 0]), linewidth=4, linecolor=:red, label = "" )
plot3d!( ([0; 0],[0; 0],[-1; 1]), linewidth=4, linecolor=:red, label = "" )

display(p)
```

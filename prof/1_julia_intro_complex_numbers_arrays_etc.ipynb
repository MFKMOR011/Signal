## Complex numbers


```julia
z = 2 + 3im 
println(z)

# You can assign j as follows: either j = im  OR: j=sqrt(-1+0im) (NB "j=sqrt(-1)" gives an error)
j = im;
z = 2 + j*3   # This works, but z = 2 + 3im is probably better as it does not involve mutiplication

println("z = ",z)

println("conj(z) = ",conj(z))    # complex conjugate

println("real(z) = ",real(z))
println("imag(z) = ",imag(z))
println("abs(z) = ",abs(z))      # magnitude
println("angle(z) = ",angle(z)," rad")  # phase
```

## For Loops

```julia
println("for n = range")    
for n = 0:0.5:2    # Specify a range
    println("n = ",n)
end

println("for n = [array of things]")
for n = ["a" "b" 1 2]  # Specify an array
    println("n= ",n)
end

println("for n in [array of things]")
for n in ["a" "b" 1 2]  # Specify an array
    println("n= ",n)
end
```

```julia
# Warning :  Julia version 1.x has an (undesirable) variable-scope inconsistency in for loops.
# Case1: code at the command line "REPL" or the "Jupyter Notebook"
# Case2: code in a script (.jl file) that is called via include("file.jl") 
# For jl scripts (Case 2), variables outside defined outside the loop are not visible within the loop. 
# Thus one needs to include the "global" statement in front of a variable that requires global scope.
# The same code running in a .jl or .ipynb file does not require the "global" keyword.  
# Because the example below is in Jupyter Notebook, one can actually omit the "global" keyword.

# Example Add the numbers 1,2,3,4 using a for loop
s = 0 
for n=1:4
    global s = s+n    # NB global statement required for jl scripts
end
println("With global, after looping, s = ",s)

# The code below would not work in an inlcuded jl script (but it does run in Jupyter and at the REPL)
s = 0 
for n=1:4
    s = s+n     # <=== Runing in a julia script would cause error "s not defined"
end
println("Without global, after looping, s = ",s)
```

## While Loop

```julia
println("Testing a simple while loop")
n=1
while n<=5
    print(" ",n)   # print() prints on same line
    n=n+1
end

```

## While Loop with break out

```julia
println()
println("Testing the break command if a condition is met")
n=1
while n<=5
    println(" ",n)
    if n==3  # Check is n equals 3
        println("  Breaking out of loop at n==3")
        break
    end
    n=n+1
end
@show n
;
```

### Plotting waveforms using Plots with Plotly backend (allows zooming with mouse)
For more examples, load file: julia_plotting_with_Plots_and_Plotly_backend.ipynb
Detailed tutorial at:  https://docs.juliaplots.org/stable/tutorial/

```julia
println("Loading Plots ... takes a few seconds.")

using Plots
plotly();    # Specify Plotly backend which allows zooming with a mouse in Jupyter Notebook.
#pythonplot()

dt = 0.002  # time step
t = collect(0:dt:1);  #Define time array from 0 to 1s in steps of 0.01s

println("length(t) = ",length(t))

A = 1
f0 = 10  # 10 Hz
omega = 2*pi*f0;   # rad/s

v = A*cos.(omega*t);   # Create an array holding the sinusoid values

fig = plot(v)
xlabel!("Time in seconds");
ylabel!("sinusoid v(t)");

display( fig )
```

## Defining functions


```julia

# One line definition
f(x) = x^2

# One line definition that works if x is an array. Squares each element.
h(x) = x.^2

# Very compact one-line definition of a rect function. Value is 0 or 1.
rect(t) = (abs.(t).<=0.5)*1.0;

# Define a multiline function
function rect(t)
    r = (abs.(t).<=0.5)*1.0
    return r
end

# Define a function of two variables
function multiply(a,b)
    return a*b
end

# Show the functions in action
@show f(10)
@show h(10)
@show rect(-1:0.1:1)
@show multiply(2,3);
t=-1:0.1:1

# Plot showing both lines joining points and markers on the points
fig = plot( t, rect(t), markershape = :circle, markersize = 2, markercolor=:lightblue, markerstrokecolor=:blue)  
title!("Magnitude of FFT(v)")
display(fig)
```

## Understanding Julia's DOT NOTATION - read this

Some Julia functions only operate on scalar quantities (not vectors):   e.g. cos() sin() abs() angle()
Some Julia functions can operate on scalars or vectors e.g. real() imag()

When you define a function, you can define it to operate only on a scaler, or on a vector.

The "dot" notation tells Julia that the argument of a function is a vector (or array or range), and that the function must be applied to each element of the vector (or array or range), and that the result must be an vector (or array of values).

f(x) = x^2    # The argument x must be a scalar value not a vector (or array or range of values)

g(x) = x.^2  # The argument x can be a vector because the .^ operation will work on a vector.

t=0:1:5    # Define a range of values 1,2,3,4,5

f(t)    # gives an error because t is not a scalar

g(t)   # returns an array containing the square of each each number i.e.  returns  [g(t[1]) g(t[2]) .... g(t[n])]

f.(t)  # works because Julia has been told that t is an array (or range), and that f() must operate on each element to produce a new array i.e.   [f(t[1]) f(t[2]) .... f(t[n])]

You can define your functions to work only on a scalar, but still use them with vectors if you include a dot when calling the function

e.g.  myfunction.(x) will work if x is an array, even if myfunction(x) only works on scalars.

```julia
f(x) = x^2    # The argument x must be a scalar value not a vector (or array or range of values)
g(x) = x.^2  # The argument x can be a vector because the .^ operation will work on a vector.

t=0:1:5    # Define a range of values 1,2,3,4,5 
@show t
@show collect(t)  # collect into an array to show all values

## @show f(t)    # gives a "MethodError" because t is not a scalar

@show g(t)   # returns an array containing the square of each each number i.e.  returns  [g(t[1]) g(t[2]) .... g(t[n])]

@show f.(t)  # works because Julia has been told that t is an array (or range)

println("Note: Attempting to evaluate f(t) will give a MethodError  - try it");
```

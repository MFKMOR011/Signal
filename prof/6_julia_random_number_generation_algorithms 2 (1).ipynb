Demonstration of the implementation of:

(1) Linear Congruential Generator (LCG) Algorithm for generating uniform distribution
https://en.wikipedia.org/wiki/Linear_congruential_generator

(2) Box–Muller transform for obtaining Gaussian via transforming a pair of uniformly distributed RV.
https://en.wikipedia.org/wiki/Box%E2%80%93Muller_transform

Note: Julia provides built-in random number generator functions: rand() and randn()
      These algorithms are just for interest.

## LCG Algorithm for generating uniform variates X~U(0,1)

```julia

# LCG Algorithm 
# Code Adapted from:
# https://github.com/h-Klok/StatsWithJuliaBook/blob/master/1_chapter/lcg.jl


m = 2^32   # Sequence will repeat with period m
a = 69069; c = 1  # Values used by older glibc library

next(X) = (a*X + c) % m

# Generate N random numbers
X = 1234   # starting number

N=10^6
data = zeros(N)

for n=1:N
   data[n] = X/m
   global X = next(X)
end

# Display the first 100 numbers
#=
using PythonPlot
figure()
plot(data[1:100],".")
title("First 100 numbers")

figure()
nbins = 30
hist(data,nbins,rwidth=0.9)   # "rwidth=0.9" makes the bins more distinct
title("Histogram of N=$(N) numbers")
;
=#

using Plots
fig = Plots.plot(data[1:100], seriestype = :scatter, markersize=2, title = "First 100 numbers", label="");
Plots.display(fig);

fig = Plots.histogram(data, nbins = 30, title = "Histogram of N=$(N)", label="");
Plots.display(fig)
```

```julia
# Calculate the mean and standard deviation
using Statistics
@show mean(data)
@show std(data)
;
```

## Box–Muller transform converts Uniform to Gaussian

```julia
# Box–Muller transform for obtaining Gaussian

function Gaussian()  # returns a single variate
    u1 = rand()
    u2 = rand()
    z0 = sqrt(-2.0 * log(u1)) * cos(2*pi * u2);
	z1 = sqrt(-2.0 * log(u1)) * sin(2*pi * u2); # Not required
    return z0
end


N=10^6
data = zeros(N)

for n=1:N
   data[n] = Gaussian()
end


sleep(0.1)

#=
using PythonPlot
figure()
plot(data[1:100],".")
title("First 100 numbers")

figure()
nbins = 30
hist(data,nbins,rwidth=0.9)   # "rwidth=0.9" makes the bins more distinct
title("Histogram")
;
=#

using Plots
fig = Plots.plot(data[1:100], seriestype = :scatter, markersize=2, title = "First 100 numbers", label="");
Plots.display(fig);

fig = Plots.histogram(data, nbins = 30, title = "Histogram of N=$(N)", label="");
Plots.display(fig)
```

```julia
# Calculate the mean and standard deviation
using Statistics
@show mean(data)
@show std(data)
;
```
